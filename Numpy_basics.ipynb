# Модификація масивів. Копії

```python
a = np.random.randint(0, 20, size=(5, 4))
a
```

```python
b = a[:2, :2]
b
```

```python
b[0, 0] = -1
b
```

```python
a # як бачимо, при модифікації масиву b змінюється вихідний масив a
```

```python
c = a[:2, :2].copy() # щоб цього не відбувалося, можна створити копію та змінювати копію
c
```

```python
c[0, 0] = 0
c
```

```python
a # тоді зміна копії не зачіпає вихідний масив
```

# З'єднання масивів

```python
x = np.array([1, 2, 3])
y = -x
```

```python
z = np.concatenate([x, y])
z
```

```python
x = np.random.random(size=(3, 4))
y = np.random.random(size=(6, 4))
print(x)
print(y)
```

```python
z = np.concatenate([x, y]) # з'єднання по першій розмірності
z
```

```python
y1 = np.random.random(size=(3, 5))
print(y1)
```

```python
z1 = np.concatenate([x, y1], axis=1) # з'єднання по другій розмірності
print(z1)
```

```python
z2 = np.hstack([x, y1]) # горизонтальне об'єднання
z2
```

```python
z3 = np.vstack([x, y]) # вертикальне об'єднання
z3
```

# Логічне індексування

```python
x = np.random.normal(172.5, 5, 10)
x = np.round(x)
x
```

```python
x > 175 # логічна маска
```

```python
x[x > 175] # ті елементи масиву х, які більше 175
```

```python
mask = (x > 160) & (x < 170) # маска з використанням двох умов одночасно
# & - AND, | - OR
mask
```

```python
x[mask] # ті елементи масиву х, які задовольняють масці
```

```python
y = np.random.randint(0, 5, size=(5, 3))
y
```

```python
mask = y[:, 1] == 0 # маска з умовою: елементи другого стовпця дорівнюють 0
mask
```

```python
y[mask, :] # ті рядки з y, які задовольняють масці
```
