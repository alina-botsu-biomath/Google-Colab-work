Відсортуйте значення за зростанням. Розділіть дані на 4 рівні за величиною групи.
Створіть новий масив (копію), в якому значення з кожної групи закодовані числами -
номерами відповідних груп.

```python
x=np.sort(p)
x
```

```python
a=x[:50].copy()# 1 group
a
```

```python
b=x[50:100].copy()# 2 group
b
```

```python
c=x[100:150].copy()# 3 group
c
```

```python
d=x[150:].copy()# 4 group
d
```

```python
a[a<360573]=1
print(a)
```

```python
b[b<499244]=2
print(b)

```

```python
c[c<612334]=3
print(c)
c[c<=612334]=3
print(c)
```

```python
d[d<749673]=4
print(d)
d[(d<=749673)&]=4
print(d)
```
