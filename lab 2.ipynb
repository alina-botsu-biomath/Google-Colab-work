Відсортуйте значення зростання. Розділіть дані на три групи:
[18; x̅− σ); [x̅− σ; x̅+ σ]; (x̅+ σ; 45], где x̅–– середній вік, σ –– стандартне
відхилення. Створіть новий масив (копію), у якому значення з першої групи
закодовані числом 1, другої - числом 2, третьої - числом 3.

```python
sorted_ages = np.sort(ages)
sorted_ages
```

```python
deviation = np.std(ages)
deviation
```

Розділіть дані на три групи: [18; x̅− σ);

```python
first_group = sorted_ages[sorted_ages < median-deviation]
first_group
```

[x̅− σ; x̅+ σ]

```python
first_step = sorted_ages[sorted_ages >= median-deviation]
middle_group = first_step[first_step <= median+deviation]
middle_group
```

(x̅+ σ; 45]

```python
last_group = sorted_ages[sorted_ages > median+deviation]
last_group
```

Створіть новий масив (копію), у якому значення з першої групи закодовані числом 1, другої - числом 2, третьої - числом 3.

```python
encoded_ages = np.concatenate((np.repeat(1, len(first_group)), np.repeat(2, len(middle_group)), np.repeat(3, len(last_group))), axis=None)
encoded_ages
```

• Створіть нову матрицю того ж розміру, елементи якої обчислюються за правилом: bij = max − aij , де aij - елементи вихідної матриці, max - її максимальний елемент.

```python
new_matrix = np.array(np.max(matrix)- matrix)
new_matrix
```

• Знайдіть рядки, перший та останній елементи яких є позитивними.

```python
positive_half = new_matrix[new_matrix[:, 0] >0]
positive = positive_half[positive_half[:, -1] > 0]
positive
```

Знайдіть стовпці, середнє значення яких більше 0.5

```python
column_means = np.mean(new_matrix, axis=0)
above05 = np.where(column_means > 0.5)[0]
above = new_matrix[:, above05]
above
```

Транспонуйте матрицю та видаліть з неї перші 5 стовпців

```python
trans = np.transpose(matrix)
trans
np.delete(trans, [1, 2,3,4,5], axis=1)
```

Додайте стовпець з одиниць (ліворуч) до матриці

```python
new_trans = np.concatenate((np.full((len(trans), 1 ), 1), trans), axis=1)
new_trans
```

Додайте до матриці п'ять нульових рядків (знизу).

```python
zero_rows = np.zeros((5, len(new_trans[0])))
zero_rows
trans_zero = np.concatenate((new_trans, zero_rows), axis=0)
trans_zero
```
