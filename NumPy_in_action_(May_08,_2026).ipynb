```python
print(A)
print(B)
```

```python
C = np.concatenate([A, B])
print(C)
```

```python
D = np.concatenate([B, A])
print(D)
```

```python
E = np.random.random(size=(3, 3))
F = np.random.randint(0, 9, size=(3, 3))
G = np.concatenate([E, F])
print(G)
```

```python
H = np.concatenate([E, F], axis=1)
print(H)
```

```python
I = np.random.randint(0, 9, size=(3, 4))
J = np.random.random(size=(5, 6))
K = np.concatenate([I, J])
print(K)
```

```python
L = np.vstack([A, B])
print(L)
```

```python
M = np.vstack([B, A])
print(M)
```

# Проста аналітика для масивів

```python
path = "/content/drive/MyDrive/IT in Analytics/Python/Spring 2023/cardio_train.csv"
data = np.loadtxt(path, dtype=np.int32, delimiter=';', skiprows=1) # зчитали весь файл, окрім заголовка
```

```python
# Зчитуємо заголовок
with open(path, 'r') as f:
  header = f.readline()[:-1].split(';')
f.close()
print(header)
```

```python
data.shape
```

```python
data.dtype
```

```python
# Порахуємо вік у роках
age = data[:, 1] // 365
data[:, 1] = age
```

```python
# Визначимо максимальний, мінімальний, середній вік пацієнтів
print(np.min(data[:, 1]))
print(np.max(data[:, 1]))
print(np.round(np.mean(data[:, 1]), 2))
```

```python
min(data[:, 1])
```

```python
%timeit min(data[:, 1])
%timeit np.min(data[:, 1])
```

```python
# Знайдемо мінімум та максимум по всім колонкам
print(np.min(data, axis=0))
print(np.max(data, axis=0))
```

```python
# Який відсоток пацієнтів має ССЗ?
np.mean(data[:, -1])
```

```python
cardio = data[:, -1]
print(cardio == 1)
```

```python
# Застосовуємо логічну маску до масиву
cardio[cardio == 1]
```

```python
print(cardio[cardio == 1].shape[0] / data.shape[0])
```

```python
people_with_cardio_problems = data[cardio == 1, :] # фільтр датасету
np.round(people_with_cardio_problems[:, 1].mean(), 2) #  середній вік пацієнтів із ССЗ
```

```python
# Знайдемо кількість людей зростом між 170 та 180 см
height = data[:, 3]
height[(height >= 170) & (height <= 180)].shape[0]
```

```python
# Знайдемо пацієнтів вагою більше 100 кг
weight = data[:, 4]
overweight = data[weight > 100, :]
```

```python
np.mean(overweight[:, -1])
```

```python
overweight[:, -1].mean()
```

```python
%timeit overweight[:, -1].mean()
%timeit np.mean(overweight[:, -1])
```
