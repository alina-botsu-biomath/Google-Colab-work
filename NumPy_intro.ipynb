## Векторизовані операції

```python
from math import sin, pi
angles = [alpha for alpha in (0, pi/2, pi, 3*pi/2)]
sin_alpha = [sin(alpha) for alpha in angles]
print(angles)
print(sin_alpha)
```

```python
angles_numpy = np.array(angles)
sin_alpha_numpy = np.sin(angles_numpy)

print(angles_numpy)
print(sin_alpha_numpy)
```

```python
np.round(sin_alpha_numpy, 1)
```

```python
2 * a
```

```python
np.round(a, 2)
```

```python
np.exp(3 * a - 5)
```

## Двовимірні масиви

```python
d = np.array([[random() for j in range(4)] for i in range(5)])
print(d)
```

```python
len(d)
```

```python
d.size # кількість елементів у матриці (5*4)
```

```python
d.ndim
```

```python
d.shape
```

```python
d.shape[0] # кількість рядків
```

```python
d.shape[1] # кількість стовпців
```

```python
d[1, 1] # звернення за індексом
```

```python
d[:, 0] # доступ до стовпця
```

```python
d[3, :] # доступ до рядка
```

```python
d[:2, :2] # доступ до частини матриці
```

```python
np.linalg.det(d[:2, :2])
```

```python
d[:, 0].reshape((5, 1)) # перетворення рядка на стовпчик
```

```python
d[:, 0].T
```

```python
np.cos(d)
```

```python
e = np.random.random((4, 2))
print(e)
```

```python
np.dot(np.cos(d), e) # матричне множення
```

```python
e * e # поелементне множення матриць
```
